```python
import castle
import numpy as np
import matplotlib.pyplot as plt
```

## 2. 初始化 Analyzer 並載入視訊


```python
# 初始化 Analyzer
analyzer = castle.Analyzer()

# 定義視訊檔案路徑
from glob import glob
video_paths = glob('reaching_videos/trial-*.mp4')

# 載入視訊檔案
analyzer.add_videos(video_paths)
print(f"已載入 {len(video_paths)} 個視訊檔案")
```

```python
# 獲取種子影像
img = analyzer.get_image()

# 定義點擊列表進行 ROI 預測
# 格式: [x, y, click_type]
click_list = [
    [100, 150, "plus_click"],   # 正向點擊
    [200, 250, "plus_click"],   # 正向點擊
    [50, 100, "minus_click"],   # 負向點擊
    [300, 200, "plus_click"]    # 正向點擊
]

# 執行 ROI 預測
roi = analyzer.predict_ROI(img, click_list)

# 儲存 ROI 提示
analyzer.save_roi_prompt(dir_path='../examples/case6-analyzer-demo/roi_prompt/')

print("分析完成！ROI 提示已儲存。")
```
